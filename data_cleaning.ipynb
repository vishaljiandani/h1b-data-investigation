```python
import pandas as pd
import numpy as np
```

```python
fy2021 = pd.read_csv("TRK_13139_FY2021.csv")
fy2021
```

```python
null_percentage = fy2021.isnull().mean() * 100

print(null_percentage)
```

```python
## Removing columns with a lot of nulls and garbage values ##

fy2021.drop(columns=['bcn', 'ben_date_of_birth', 'WAGE_AMT', 'WAGE_UNIT', 'FULL_TIME_IND', 'S1Q1B',
                     'RECEIPT_NUMBER', 'T_U_VAWA_FLAG', 'DOL_ETA_CASE_NUMBER'], inplace=True)
fy2021
```

```python
fy2021.drop(fy2021[fy2021['ben_multi_reg_ind'] == '(b)(3) (b)(6) (b)(7)(c)'].index, inplace=True)
```

```python
fy2021 = fy2021.astype({
    'NUM_OF_EMP_IN_US': 'float',
    'ben_year_of_birth': 'int',
    'lottery_year': 'int',
    'ben_multi_reg_ind': 'int',
    'rec_date': 'datetime64[ns]',
    'first_decision_date': 'datetime64[ns]',
    'valid_from': 'datetime64[ns]',
    'valid_to': 'datetime64[ns]',
    'BEN_COMP_PAID': 'float'
})
```
